```python
numerical_clm = ['PM2.5','PM10', 'NO','NO2', 'NOx', 'NH3', 'CO', 'SO2','O3','Benzene', 'Toluene']
```

```python
for clm in numerical_clm:
    plt.figure(figsize= (6,4))
    sns.boxplot(data = df, x = clm)
```

```python
sns.heatmap(df[numerical_clm].corr(), annot=True, cmap="coolwarm")
```

```python
X = df.drop(columns= 'AQI_Bucket')
y = df['AQI_Bucket']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    random_state= 42, 
    test_size= 0.2,
    stratify= y
)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy= 'median')
```

```python
X_train[numerical_clm] = imputer.fit_transform(
    X_train[numerical_clm]
)
```

```python
X_test[numerical_clm] = imputer.transform(
    X_test[numerical_clm]
)
```

```python
X_test[numerical_clm].isnull().sum()
```

```python
X_train = pd.get_dummies(X_train, columns=['City'])
X_test = pd.get_dummies(X_test, columns=['City'])
```

```python
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
```

```python
X_train = X_train.astype(int)
X_test = X_test.astype(int)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Logistic regression model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model_lr = LogisticRegression(random_state= 42,
                              max_iter= 1000)
```

```python
model_lr.fit(X_train_scaled, y_train)
```

```python
y_pred_lr = model_lr.predict(X_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score,classification_report
```

```python
accuracy_score(y_pred_lr, y_test)
```

```python
print(classification_report(y_pred_lr, y_test))
```

```python
cm = confusion_matrix(y_test, y_pred_lr)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

Random forest model

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf = RandomForestClassifier(
    n_estimators= 200,
    random_state= 42
)
```

```python
model_rf.fit(X_train,y_train)
```

```python
y_pred_rf = model_rf.predict(X_test)
```

```python
accuracy_score(y_pred_rf, y_test)
```

```python
print(classification_report(y_pred_rf, y_test))
```

```python
cm2 = confusion_matrix(y_pred_rf, y_test)
sns.heatmap(cm2, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Random Forest - Confusion Matrix")
plt.show()
```
