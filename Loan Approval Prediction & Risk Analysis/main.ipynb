```python
sns.countplot(x = df['loan_status'])
plt.title('Loan ammount count plot')
```

```python
numeric_clm = ['person_age','person_income',
       'person_emp_exp', 'loan_amnt',
       'loan_int_rate', 'loan_percent_income', 'cb_person_cred_hist_length',
       'credit_score']
```

```python
for clm in numeric_clm:
    plt.figure(figsize= (6,4))
    sns.histplot(data= df,x= clm, kde= True)
    plt.title(f'Distribution of {clm}')
```

```python
outlier_clm = ['person_age', 'person_income', 'person_emp_exp', 'loan_amnt']
```

```python
for clm in outlier_clm:
    plt.figure(figsize= (6,4))
    sns.boxplot(x= df[clm])
```

```python
categoic_clm = ['person_gender', 'person_education',
       'person_home_ownership','loan_intent',
        'previous_loan_defaults_on_file']
```

```python
for clm in categoic_clm:
    plt.figure(figsize= (6,4))
    sns.countplot(data= df, x = clm)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_clm].corr(), cmap ='coolwarm', annot= True)
plt.show()
```

```python
for clm in ['credit_score','person_income','loan_amnt','loan_int_rate','loan_percent_income']:
    plt.figure(figsize=(6,4))
    sns.boxplot(data = df,x =  'loan_status', y =clm)
```
